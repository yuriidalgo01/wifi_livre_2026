```python
# Lendo os arquivos dos Países do Mundo:

# Importando o geopandas

import geopandas as gpd
import pathlib as Path
```

```python
# Varegando dos dados

path = (r"C:\Users\x432601\Documents\GitHub\ETL_wifi_livre_2026\data\ne_10m_admin_0_countries.zip")

gdf_countries = gpd.read_file(path)
```

```python
# Verificando a tabela

gdf_countries.head()
```

```python
# Plotando o mapa dos países do mundo

gdf_countries.plot()
```

```python
# Calculando centróides dos países

# Reprojeta para CRS projetado em metros, calcula o centroide e converte de volta ao CRS original
gdf_countries['centroid'] = (
    gdf_countries.geometry
    .to_crs(epsg=3857)       # Projeta para metros (Web Mercator)
    .centroid
    .to_crs(gdf_countries.crs)  # Retorna para o CRS original (graus)
)
```

```python
# Verificando os dados com os centróides calculados

gdf_countries.head()
```

```python
#print("Colunas do GeoDataFrame:", gdf_countries.columns)
print("[")
for elem in gdf_countries.columns:
    print(f"{elem},")
print("]")
```

```python
# Selecionando apenas as colunas relevantes para o projeto
gdf_countries_clean = gdf_countries[['ADMIN', 'geometry', 'centroid']]
```

```python
# Plot centroides dos países do mundo
# Plot demora para ser carregado, e está errado. Deveria plotar os centróides, mas está plotando os polígonos dos países pintados.
ax = gdf_countries_clean.set_geometry('centroid').plot("ADMIN", markersize = 5)

#Plot polygons com prenchimento de cor

gdf_countries_clean.plot("ADMIN",
                         ax = ax, facecolor = "none", 
                         edgecolor = "k", linewidth = 0.2)
```
